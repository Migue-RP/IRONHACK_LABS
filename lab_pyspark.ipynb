```python
rdd = df.rdd
```

```python
print(type(rdd))
```

```python
rdd.take(5)
```

```python
name_email_rdd = rdd.map(lambda x: (x["CUSTOMERNAME"], x["EMAIL"]))

name_email_rdd.take(5)
```

```python
age_above_35 = rdd.filter(lambda x: x["age"] > 35)

age_above_35.take(5)
```

```python
domains_rdd = rdd.map(lambda x: x["EMAIL"].split("@")[1])

domains_rdd.take(5)
```

```python
domain_count = domains_rdd.map(lambda x: (x, 1)) \
                          .reduceByKey(lambda a, b: a + b)

domain_count.collect()
```

```python
customer_country_rdd = rdd.map(lambda x: (x["d"], x["COUNTRY"]))

customer_country_rdd.take(5)
```

```python
cust_id_domain_rdd = rdd.map(lambda x: (x["d"], x["EMAIL"].split("@")[1]))
```

```python
cust_id_domain_df = cust_id_domain_rdd.toDF(["CustomerID", "EmailDomain"])
```

```python
joined_df = df.join(cust_id_domain_df,
    df["d"] == cust_id_domain_df["CustomerID"],
    "inner")
```

```python
joined_df.select("d", "EMAIL", "EmailDomain").show()
```
