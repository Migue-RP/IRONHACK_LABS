```python
df = spark.read.table("spotify_data")
df.show(10)
```

```python

df.select("song_title", "duration_ms").sort("duration_ms", ascending=False).first()
```

```python
df.filter(df["artist"] == "Drake").count()
```

```python
from pyspark.sql.functions import col
df.select("song_title").filter(col("song_title").contains("Love")).show()
```

```python
df.select("song_title").filter(col("song_title").rlike("[0-9]")).show()
```

```python
song_bachelorette = df.select("song_title").sort("energy", ascending=False).limit(20)
song_bachelorette.show()
```

```python
dance_songs = df.select("song_title").sort("danceability", ascending=False).limit(3)
dance_songs.show()
```

```python
from pyspark.sql.functions import count
df.groupBy("artist").agg(count("song_title").alias("song_count")).orderBy(col("song_count").desc()).first()
```

```python
break_up_list = df.select("song_title").filter((col("instrumentalness") > 0.8) & (col("liveness") > 0.2))
break_up_list.show()
```
